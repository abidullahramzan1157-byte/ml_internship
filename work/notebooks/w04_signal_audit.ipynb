## 2. Signal test #1 / #2 / #3 (verdict each)

*Three safe signals, each with a mini-test and a verdict: CONFIRMED / OPPOSITE / MIXED / FALSE.*

```python
import pandas as pd

df = pd.read_csv("house_prices_prediction.csv")

# Signal 1: Larger area → higher price
median_area = df["Area_sqft"].median()

large_area = df[df["Area_sqft"] >= median_area]["Price_PKR"].mean()
small_area = df[df["Area_sqft"] < median_area]["Price_PKR"].mean()

print("Signal #1: Area")
print("Large-area average price:", large_area)
print("Small-area average price:", small_area)

if large_area > small_area:
    print("Verdict: CONFIRMED")
else:
    print("Verdict: OPPOSITE")


# Signal 2: More bedrooms → higher price
median_bedrooms = df["Bedrooms"].median()

more_bedrooms = df[df["Bedrooms"] >= median_bedrooms]["Price_PKR"].mean()
fewer_bedrooms = df[df["Bedrooms"] < median_bedrooms]["Price_PKR"].mean()

print("\nSignal #2: Bedrooms")
print("More-bedroom average price:", more_bedrooms)
print("Fewer-bedroom average price:", fewer_bedrooms)

if more_bedrooms > fewer_bedrooms:
    print("Verdict: CONFIRMED")
else:
    print("Verdict: OPPOSITE")


# Signal 3: Newer houses → higher price
median_age = df["Age_years"].median()

newer = df[df["Age_years"] < median_age]["Price_PKR"].mean()
older = df[df["Age_years"] >= median_age]["Price_PKR"].mean()

print("\nSignal #3: Age")
print("Newer-house average price:", newer)
print("Older-house average price:", older)

if newer > older:
    print("Verdict: CONFIRMED")
else:
    print("Verdict: OPPOSITE")
```

| Signal          | Mini-test                                                   | Verdict                                         |
| --------------- | ----------------------------------------------------------- | ----------------------------------------------- |
| **#1 Area**     | Compare average price of larger vs smaller houses           | **CONFIRMED** if larger-area average is higher  |
| **#2 Bedrooms** | Compare average price of houses with more vs fewer bedrooms | **CONFIRMED** if more-bedroom average is higher |
| **#3 Age**      | Compare average price of newer vs older houses              | **CONFIRMED** if newer-house average is higher  |
