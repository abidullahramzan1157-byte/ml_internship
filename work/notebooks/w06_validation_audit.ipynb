## 2. My model under an honest split (before/after)

*Re-run your Week-5 model under a grouped or time-aware split. Show both numbers.*

My Model Under an Honest Split (Before/After)

Because your dataset has no client ID or date/time column, a true grouped or time-aware split is not possible. The honest approach is to state that limitation rather than invent a grouping.

Use this code to show the before (random split) and after (honest limitation):

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

df = pd.read_csv("house_prices_prediction.csv")

X = df[["Area_sqft", "Bedrooms", "Bathrooms", "Age_years"]]
y = df["Price_PKR"]

# BEFORE: Week-5 random 80/20 split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)

pred = model.predict(X_test)
before_mae = mean_absolute_error(y_test, pred)

print("Before — Random split MAE:", before_mae)

# AFTER:
# No client ID or time column exists, so a grouped/time-aware
# split cannot be performed honestly.

print("After — Grouped/time-aware MAE: Not applicable")
print("Reason: Dataset has no client/group or time column.")
```
