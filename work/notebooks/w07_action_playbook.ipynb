### 3. Human Review + the No-Go List

**What a person must check before acting:**

* Verify the house details, such as area, bedrooms, bathrooms, and age.
* Check whether the data is correct and complete.

* Review unusual or extreme cases manually.
* Compare the model estimate with real market information before making a decision.

* Final buying or selling decisions.

### 5. Exports for the Paper

I exported the ranked queue and supporting figures to the `work/outputs/` folder so they can be reused in the final paper.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Create output folder
os.makedirs("work/outputs", exist_ok=True)

# Load dataset
df = pd.read_csv("house_prices_prediction.csv")

# -----------------------------
# 1. Create ranked queue
# -----------------------------
df["score"] = (
    df["Area_sqft"] * 0.4 +
    df["Bedrooms"] * 0.2 +
    df["Bathrooms"] * 0.2 -
    df["Age_years"] * 0.2
)

df["rank"] = (
    df["score"]
    .rank(ascending=False, method="first")
    .astype(int)
)

df = df.sort_values("rank")

df.to_csv(
    "work/outputs/baseline_action_score.csv",
    index=False
)

# -----------------------------
# 2. Export top 20
# -----------------------------
top20 = df.head(20).copy()

top20["action"] = "Review for high-value house"
top20["reason_code"] = "HIGH_VALUE"

top20.to_csv(
    "work/outputs/top20_review.csv",
    index=False
)

# -----------------------------
# 3. Price distribution figure
# -----------------------------
plt.figure(figsize=(8, 5))
plt.hist(df["Price_PKR"], bins=20)
plt.xlabel("Price (PKR)")
plt.ylabel("Number of Houses")
plt.title("House Price Distribution")
plt.tight_layout()
plt.savefig(
    "work/outputs/price_distribution.png",
    dpi=300
)
plt.show()

# -----------------------------
# 4. Feature distributions
# -----------------------------
features = [
    "Area_sqft",
    "Bedrooms",
    "Bathrooms",
    "Age_years"
]

for feature in features:
    plt.figure(figsize=(7, 4))
    plt.hist(df[feature], bins=20)
    plt.xlabel(feature)
    plt.ylabel("Number of Houses")
    plt.title(f"{feature} Distribution")
    plt.tight_layout()
    plt.savefig(
        f"work/outputs/{feature}_distribution.png",
        dpi=300
    )
    plt.show()

print("All exports saved in work/outputs/")
```
